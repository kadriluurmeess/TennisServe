## a. Andmestiku kirjeldus

WTA üksikmängud aastatel 2018–2023. Allikas: [TennisMyLife](https://stats.tennismylife.org/tennis-match-database).

Käivita märkmik repositooriumi kaustas. Kui paketid puuduvad, paigalda need terminalis käsuga `python -m pip install -r requirements.txt`. Algandmed laaditakse kohalikku `data/raw/` kausta.


```python
import pandas as pd
from download_data import download_data

# Olemasolevad aastafailid säilitatakse; puuduvad laaditakse algallikast.
failid = download_data()
```

```python
# Loeme iga aasta eraldi ja ühendame failid ühtseks tabeliks.
aastatabelid = [pd.read_csv(fail) for fail in failid]
df = pd.concat(aastatabelid, ignore_index=True)

print(f"Ridade arv: {df.shape[0]}")
print(f"Veergude arv: {df.shape[1]}")

ulevaade = pd.DataFrame([
    {"Aasta": int(fail.name[:4]), "Ridu": tabel.shape[0], "Veerge": tabel.shape[1]}
    for fail, tabel in zip(failid, aastatabelid)
])
ulevaade
```
